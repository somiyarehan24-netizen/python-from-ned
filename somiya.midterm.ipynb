```python
# Write a program to manage student marks using lists and
#dictionaries.
#i #Input details of 5 students (name and marks).
# ii.#Store them in a dictionary.
# iii.#print the list using for loop
# iv. #Display highest and lowest marks. (use max and min methods)
#ans: 
 #i) Input details of 5 students (name and marks).
name1 = input("Enter name of student 1 : ")
marks1 = float(input("Enter your Marks :"))
 
name2 = input("Enter name of student 2  : ")
marks2 = float(input("Enter your Marks :"))
 
name3 = input("Enter name of student 3 :")
marks3 = float(input("Enter your Marks :"))
 
name4 = input("Enter name of student 4 : ")
marks4 = float(input("Enter your Marks :"))
 
name5 = input("Enter name of student 5 :")
marks5 = float(input("Enter your Marks :"))

#ii) Store them in a dictionary.
Students = {
    name1: marks1,
    name2: marks2,
    name3: marks3,
    name4: marks4,
    name5: marks5
}
print(Students)

 #iii)print the list using for loop
for name, marks in Students.items():
    print(name,marks)

 #iv) Display highest and lowest marks. (use max and min methods)
marks = [85, 92, 78, 64, 99, 88, 71]

highest_mark = max(marks)
lowest_mark = min(marks)

print(f"Highest Mark: {highest_mark}")
print(f"Lowest Mark: {lowest_mark}")
```

```python
 A class has students enrolled in different clubs.
#i Create a nested dictionary where each club stores a list of student
names
#2 .Create 2 new sets having club members
#iii Find students common to two clubs using sets made above in last part.
#iv Ask the user for a club name and display its members using conditional statment
answer:
# i: Nested dictionary with list of members.
clubs = {
    "Music": ["somiya", "ali", "amna"],
    "Sports": ["sara", "ayesha", "ruhail"]
}

#ii: Create 2 sets from the clubs.
music_set = set(clubs["Music"])
sports_set = set(clubs["Sports"])

print("Music Members:", music_set)
print("Sports Members:", sports_set)

#iii: Find common students using sets
common_students = music_set & sports_set
print("Common Members:", common_students)

#iv: Ask user for a club name and display members
user_input = input("Enter club name (Music or Sports): ")

if user_input in clubs:
    print("Members:", clubs[user_input])
else:
    print("Club not found!")
```
