```python
Q2 #Create a dictionary which has questions as key and answers as the value to
#use in this question.
#Part A — Conduct the Quiz
```

```python
import random
```

```python
questions = {
    "What is the capital of France?": "Paris",
    "How many days are in a standard week?": "7",
    "Which language am i learning?": "Python",
    "What comes after 9?": "10",
    "how many colours are in ranbow?": "7",
}

def start_quiz(questions):
    selected_questions = random.sample(list(questions.keys()), 3)
    score = 0

    for question in selected_questions:
        print(question)
        answer = input("Enter your answer: ")

        if answer.strip().lower() == questions[question].lower():
            print("Correct!")
            score += 1
        else:
            print("Incorrect!")
            print("Correct answer:", questions[question])

    print("Your final score is:", score, "out of 3")
    return score

start_quiz(questions)
```

```python
#Part B — Generate a Quiz Report (Continuation of Q2)
#Write a function generate_report(name, score) that:
#1. Uses the datetime module to get the current date and time when
#the report is generated. (1 mark)
#2. Calculates the percentage score out of 3 and assigns a result:
#o 100%: Excellent
#o 67% or 33%: Good
# o 0%: Needs Improvement (2 marks)
#3. Stores the student's name, score, percentage, result, and date/time in
#a dictionary and displays the report neatly. (2 marks)
```

```python
from datetime import datetime

def generate_report(name, score):
    percentage = (score / 3) * 100

    if percentage == 100:
        result = "Excellent or Marvellous"
    elif percentage >= 33:
        result = "Good"
    else:
        result = "Needs much Improvement"

    current_time = datetime.now()

    report = {
        "Student": name,
        "Score": f"{score}/3",
        "Percentage": f"{percentage:.2f}%",
        "Result": result,
        "Date and Time": current_time
    }

    print("--- QUIZ REPORT ---")
    for key, value in report.items():
        print(f"{key}: {value}")


generate_report("Ali", 1)
```

```python
students = {
    "safeer": [85, 90, 80],
    "hamza": [95, 92, 88],
    "Moiz": [65, 70, 60],
    "sami": [75, 80, 85],
    "Somiya": [50, 55, 45]
}
```

```python
#b) calculate Student Results
#Write a function calculate_results(students) that:
# Uses a loop to calculate each student's average marks.
# Assigns a grade based on their average:
#o 90 or above: A
#o 80–89: B
#o 70–79: C
#o 60–69: D
#o Below 60: F
# Returns a new dictionary containing each student's average and grade.
```

```python
def calculate_results(students):
    results = {}

    for name, marks in students.items():
        total = 0

        for mark in marks:
            total += mark

        average = total / len(marks)

        if average >= 90:
            grade = "A"
        elif average >= 80:
            grade = "B"
        elif average >= 70:
            grade = "C"
        elif average >= 60:
            grade = "D"
        else:
            grade = "Fail"

        results[name] = {
            "Average": average,
            "Grade": grade
        }

    return results
```

```python
#c) Find Top Performers
#Write a function find_top_students(students) that:
# Calculates each student's average using a loop.
# Finds and returns the names of all students whose average is 80 or higher.
# Stores the qualifying names in a list.
# Does not use any built-in sorting functions.
```

```python
def find_top_students(students):
    top_students = []

    for name, marks in students.items():
        total = 0

        for mark in marks:
            total += mark

        average = total / len(marks)

        if average >= 80:
            top_students.append(name)

    return top_students


print("Student Results:")
print(calculate_results(students))

print("\nTop Students:")
print(find_top_students(students))
```

```python
#Design a Python program for a Library Book Management System using
#classes, objects, and inheritance.
#Create the following classes:

#A. Class Book (4 Marks)

#● Define attributes: title, author, and price.
#● Create a constructor to initialize these attributes.
#● Define a method display_info() to display the book details.
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_info(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Price:", self.price)
```

```python
#part B
class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    def display_info(self):
        super().display_info()
        print("File Size:", self.file_size, "MB")
```

```python
#part C
book1 = Book("Accounting", "Somiya", 2000)
ebook1 = EBook("Data Science", "Safeer Ahmed", 8000, 5)

print("--- BOOK DETAILS ---")
book1.display_info()

print("\n--- EBOOK DETAILS ---")
ebook1.display_info()

```
