```python
alphabet = str(input("enter your character"))
if alphabet == "a":
    print("alphabet is vowel")
elif alphabet == "e":
    print("alphabet is vowel")
elif alphabet == "i":
    print("alphabet is vowel")
elif alphabet == "o":
    print("alphabet is vowel")
elif alphabet == "u":
    print("alphabet is vowel")
else:
    print("alphabet is consonent")
```

```python
marks = int(input("enter your marks"))
if marks >= 90 and marks <= 100:
    print("grade is A1 ")
elif marks >= 80 and marks < 90:
    print(" grade is A")
elif marks >= 70 and marks <80:
    print(" grade is B ")
elif marks >= 60 and marks <70:
    print(" grade is C ")
elif marks >= 50 and marks <60:
    print(" grade is D")
else:
    print("fail")


```

```python
year = int(input("enter the year"))
if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print("year is leap")
        else:
            print("year is not leap")
    else:
        print("year is leap")
else:
    print("year is not leap")
```
